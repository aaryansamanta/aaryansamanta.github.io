```python
# CAPSTONE PROJECT: DIABETES PREDICTION WITH MACHINE LEARNING

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

# 2. UPLOAD DATASET
print("=" * 70)
print("UPLOAD diabetes.csv")
print("=" * 70)

uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print(f"File: {filename}")


# 3. INITIAL DATA EXPLORATION

print("\n" + "=" * 70)
print("STEP 1: DATASET EXPLORATION")
print("=" * 70)

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nDataset shape:")
print(df.shape)

print("\nMissing values BEFORE cleaning:")
print(df.isnull().sum())

print("\nStatistical summary:")
display(df.describe())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# 4. DATA CLEANING

print("\n" + "=" * 70)
print("STEP 2: DATA CLEANING & PREPROCESSING")
print("=" * 70)

# In the Pima dataset, zero is biologically implausible/missing
# for these variables.
zero_as_missing = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

df_clean = df.copy()

print("\nZero values BEFORE replacement:")
for col in zero_as_missing:
    print(f"{col}: {(df_clean[col] == 0).sum()}")

# Replace zeros with NaN
df_clean[zero_as_missing] = df_clean[zero_as_missing].replace(0, np.nan)

print("\nMissing values AFTER replacing invalid zeros:")
print(df_clean.isnull().sum())

# Separate features and target
X = df_clean.drop("Outcome", axis=1)
y = df_clean["Outcome"]

feature_names = X.columns.tolist()

# Train/test split BEFORE fitting imputer/scaler to avoid data leakage
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# 5. IMPUTATION

imputer = SimpleImputer(strategy="mean")

X_train_imputed = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=feature_names,
    index=X_train.index
)

X_test_imputed = pd.DataFrame(
    imputer.transform(X_test),
    columns=feature_names,
    index=X_test.index
)

print("\nMissing values AFTER mean imputation:")
print(X_train_imputed.isnull().sum())


# 6. EDA VISUALIZATIONS

print("\n" + "=" * 70)
print("STEP 3: EXPLORATORY DATA ANALYSIS")
print("=" * 70)

# Visualization 1: Age Distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df_clean, x="Age", bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()

print("Insight: The dataset contains patients across a broad range of ages,")
print("with a larger concentration among younger and middle-aged adults.")

# Visualization 2: Pregnancy Count
plt.figure(figsize=(10, 5))
sns.histplot(data=df_clean, x="Pregnancies", discrete=True)
plt.title("Pregnancy Count Distribution")
plt.xlabel("Number of Pregnancies")
plt.ylabel("Number of Patients")
plt.show()

print("Insight: Most patients have relatively few pregnancies, while a smaller")
print("number of patients have substantially higher pregnancy counts.")

# Visualization 3: Glucose Levels
plt.figure(figsize=(8, 5))
sns.histplot(data=df_clean, x="Glucose", bins=25, kde=True)
plt.title("Glucose Level Distribution")
plt.xlabel("Glucose")
plt.ylabel("Number of Patients")
plt.show()

print("Insight: Glucose levels vary considerably. Higher glucose values are")
print("generally associated with greater diabetes risk.")

# Visualization 4: BMI Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_clean, x="BMI")
plt.title("BMI Distribution and Outliers")
plt.xlabel("BMI")
plt.show()

print("Insight: BMI contains variation and some potentially extreme values.")
print("Higher BMI is generally associated with increased diabetes risk.")

# Visualization 5: Outcome Count
plt.figure(figsize=(7, 5))
sns.countplot(data=df_clean, x="Outcome")
plt.title("Diabetes Outcome Count")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")
plt.show()

print("Insight: The dataset contains more non-diabetic patients than diabetic")
print("patients, so the classes are not perfectly balanced.")

# Additional useful visualization: correlation heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    df_clean.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Feature Correlation Heatmap")
plt.show()

print("Insight: Glucose generally shows one of the strongest relationships")
print("with diabetes outcome, while several health indicators also show")
print("meaningful relationships.")


# 7. MODEL 1: LOGISTIC REGRESSION

print("\n" + "=" * 70)
print("STEP 4: MACHINE LEARNING MODELS")
print("=" * 70)

logistic_model = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000, random_state=42))
])

logistic_model.fit(X_train, y_train)
logistic_pred = logistic_model.predict(X_test)


# 8. MODEL 2: RANDOM FOREST

rf_model = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("model", RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ))
])

rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)


# 9. MODEL 3: SVM

svm_model = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()),
    ("model", SVC(random_state=42))
])

svm_model.fit(X_train, y_train)
svm_pred = svm_model.predict(X_test)


# 10. MODEL EVALUATION FUNCTION

def evaluate_model(name, y_true, predictions):
    accuracy = accuracy_score(y_true, predictions)
    precision = precision_score(y_true, predictions, zero_division=0)
    recall = recall_score(y_true, predictions, zero_division=0)
    f1 = f1_score(y_true, predictions, zero_division=0)

    print("\n" + "-" * 60)
    print(name)
    print("-" * 60)

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1 Score:  {f1:.4f}")

    print("\nClassification Report:")
    print(classification_report(
        y_true,
        predictions,
        target_names=["No Diabetes", "Diabetes"],
        zero_division=0
    ))

    print("Confusion Matrix:")
    print(confusion_matrix(y_true, predictions))

    return {
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    }

results = []

results.append(
    evaluate_model("Logistic Regression", y_test, logistic_pred)
)

results.append(
    evaluate_model("Random Forest", y_test, rf_pred)
)

results.append(
    evaluate_model("Support Vector Machine", y_test, svm_pred)
)


# 11. MODEL COMPARISON TABLE

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)


# 12. CONFUSION MATRIX

plt.figure(figsize=(7, 6))

cm = confusion_matrix(y_test, rf_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Diabetes", "Diabetes"],
    yticklabels=["No Diabetes", "Diabetes"]
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()


# 13. RANDOM FOREST HYPERPARAMETER TUNING

print("\n" + "=" * 70)
print("STEP 5: RANDOM FOREST HYPERPARAMETER TUNING")
print("=" * 70)

rf_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("model", RandomForestClassifier(random_state=42))
])

param_grid = {
    "model__n_estimators": [100, 200, 300],
    "model__max_depth": [None, 5, 10, 15],
    "model__min_samples_split": [2, 5, 10]
}

grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)

best_rf_model = grid_search.best_estimator_

print("\nBest Parameters:")
print(grid_search.best_params_)

print(f"\nBest Cross-Validation Accuracy: {grid_search.best_score_:.4f}")

# Predictions from tuned model
tuned_rf_pred = best_rf_model.predict(X_test)

# Evaluate tuned RF
tuned_accuracy = accuracy_score(y_test, tuned_rf_pred)
tuned_precision = precision_score(
    y_test, tuned_rf_pred, zero_division=0
)
tuned_recall = recall_score(
    y_test, tuned_rf_pred, zero_division=0
)
tuned_f1 = f1_score(
    y_test, tuned_rf_pred, zero_division=0
)

print("\nTUNED RANDOM FOREST RESULTS")
print("-" * 60)
print(f"Accuracy:  {tuned_accuracy:.4f}")
print(f"Precision: {tuned_precision:.4f}")
print(f"Recall:    {tuned_recall:.4f}")
print(f"F1 Score:  {tuned_f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    tuned_rf_pred,
    target_names=["No Diabetes", "Diabetes"],
    zero_division=0
))


# 14. BEFORE VS AFTER TUNING

original_rf_accuracy = accuracy_score(y_test, rf_pred)

before_after_df = pd.DataFrame({
    "Model": ["Random Forest Before Tuning", "Random Forest After Tuning"],
    "Accuracy": [original_rf_accuracy, tuned_accuracy],
    "Precision": [
        precision_score(y_test, rf_pred, zero_division=0),
        tuned_precision
    ],
    "Recall": [
        recall_score(y_test, rf_pred, zero_division=0),
        tuned_recall
    ],
    "F1 Score": [
        f1_score(y_test, rf_pred, zero_division=0),
        tuned_f1
    ]
})

print("\n" + "=" * 70)
print("BEFORE VS AFTER HYPERPARAMETER TUNING")
print("=" * 70)

display(
    before_after_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

# Accuracy comparison chart
plt.figure(figsize=(9, 5))

sns.barplot(
    data=before_after_df,
    x="Model",
    y="Accuracy"
)

plt.ylim(0, 1)
plt.title("Random Forest Accuracy: Before vs. After Tuning")
plt.ylabel("Accuracy")
plt.xlabel("")
plt.xticks(rotation=15)
plt.show()


# 15. FEATURE IMPORTANCE

print("\n" + "=" * 70)
print("STEP 6: FEATURE IMPORTANCE ANALYSIS")
print("=" * 70)

# Extract Random Forest inside the pipeline
rf_estimator = best_rf_model.named_steps["model"]

# Get feature importances
importance_values = rf_estimator.feature_importances_

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance_values
}).sort_values(
    by="Importance",
    ascending=False
)

print("\nAll feature importance scores:")
display(
    importance_df.style.format({
        "Importance": "{:.4f}"
    })
)

top5 = importance_df.head(5)

print("\nTop 5 Most Important Features:")
display(
    top5.style.format({
        "Importance": "{:.4f}"
    })
)

plt.figure(figsize=(9, 6))

sns.barplot(
    data=top5,
    x="Importance",
    y="Feature"
)

plt.title("Top 5 Most Important Features")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.show()

print("\nFeature Importance Analysis:")
for i, row in top5.iterrows():
    print(
        f"- {row['Feature']}: importance = {row['Importance']:.4f}"
    )

print(
    "\nThese features were considered most useful by the Random Forest "
    "for separating patients with and without diabetes. Feature importance "
    "shows how useful a feature was to the model, but it does not prove "
    "that the feature directly causes diabetes."
)


# 16. SELECT BEST MODEL

print("\n" + "=" * 70)
print("SELECTING BEST MODEL")
print("=" * 70)

# Compare tuned RF with original models
all_model_results = pd.DataFrame([
    {
        "Model": "Logistic Regression",
        "Accuracy": accuracy_score(y_test, logistic_pred),
        "Precision": precision_score(y_test, logistic_pred, zero_division=0),
        "Recall": recall_score(y_test, logistic_pred, zero_division=0),
        "F1 Score": f1_score(y_test, logistic_pred, zero_division=0)
    },
    {
        "Model": "Random Forest",
        "Accuracy": accuracy_score(y_test, rf_pred),
        "Precision": precision_score(y_test, rf_pred, zero_division=0),
        "Recall": recall_score(y_test, rf_pred, zero_division=0),
        "F1 Score": f1_score(y_test, rf_pred, zero_division=0)
    },
    {
        "Model": "SVM",
        "Accuracy": accuracy_score(y_test, svm_pred),
        "Precision": precision_score(y_test, svm_pred, zero_division=0),
        "Recall": recall_score(y_test, svm_pred, zero_division=0),
        "F1 Score": f1_score(y_test, svm_pred, zero_division=0)
    },
    {
        "Model": "Tuned Random Forest",
        "Accuracy": tuned_accuracy,
        "Precision": tuned_precision,
        "Recall": tuned_recall,
        "F1 Score": tuned_f1
    }
])

display(
    all_model_results.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

best_model_name = all_model_results.loc[
    all_model_results["Accuracy"].idxmax(),
    "Model"
]

print(f"\nBest model based on test accuracy: {best_model_name}")

# Use tuned RF as the final model if it performs best;
# otherwise use whichever model achieved the highest accuracy.
if best_model_name == "Tuned Random Forest":
    final_model = best_rf_model
elif best_model_name == "Random Forest":
    final_model = rf_model
elif best_model_name == "Logistic Regression":
    final_model = logistic_model
else:
    final_model = svm_model


# 17. NEW PATIENT PREDICTION FUNCTION

print("\n" + "=" * 70)
print("STEP 7: NEW PATIENT PREDICTIONS")
print("=" * 70)

def predict_diabetes(model, patient_data, patient_name="Patient"):
    """
    Predict diabetes risk for a new patient.
    Uses the preprocessing stored inside the trained model pipeline.
    """

    patient_df = pd.DataFrame([patient_data])

    # Ensure the exact feature order
    patient_df = patient_df[feature_names]

    prediction = model.predict(patient_df)[0]

    # Probability if model supports predict_proba
    probability = None

    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(patient_df)[0][1]

    elif hasattr(model, "named_steps"):
        final_estimator = model.named_steps["model"]

        if hasattr(final_estimator, "predict_proba"):
            probability = model.predict_proba(patient_df)[0][1]

    print(f"\n{patient_name}")
    print("-" * 50)
    print("Patient data:")
    display(patient_df)

    if probability is not None:
        print(f"Estimated diabetes probability: {probability:.2%}")

    if prediction == 1:
        print("🔴 HIGH RISK: Model predicts diabetes.")
    else:
        print("🟢 LOW RISK: Model predicts no diabetes.")

    return prediction, probability


# 18. HIGH-RISK PATIENT

high_risk_patient = {
    "Pregnancies": 8,
    "Glucose": 180,
    "BloodPressure": 90,
    "SkinThickness": 35,
    "Insulin": 200,
    "BMI": 36.0,
    "DiabetesPedigreeFunction": 1.0,
    "Age": 55
}

high_prediction, high_probability = predict_diabetes(
    final_model,
    high_risk_patient,
    "Patient #1 — High-Risk Profile"
)


# 19. LOW-RISK PATIENT

low_risk_patient = {
    "Pregnancies": 0,
    "Glucose": 90,
    "BloodPressure": 70,
    "SkinThickness": 20,
    "Insulin": 80,
    "BMI": 21.5,
    "DiabetesPedigreeFunction": 0.2,
    "Age": 25
}

low_prediction, low_probability = predict_diabetes(
    final_model,
    low_risk_patient,
    "Patient #2 — Low-Risk Profile"
)


# 20. FINAL SUMMARY

print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

print(f"""
Dataset:
- Number of patients: {df.shape[0]}
- Number of original features: {df.shape[1] - 1}
- Target variable: Outcome

Preprocessing:
- Invalid zero values were treated as missing for relevant health variables.
- Missing values were replaced using mean imputation.
- Data was split into 80% training and 20% testing.
- StandardScaler was used for Logistic Regression and SVM.
- Random Forest was not scaled because tree-based models do not require feature scaling.

Models:
1. Logistic Regression
2. Random Forest
3. Support Vector Machine (SVM)

Hyperparameter Tuning:
- GridSearchCV was used for Random Forest.
- Parameters tested included:
  * n_estimators
  * max_depth
  * min_samples_split

Best Random Forest Parameters:
{grid_search.best_params_}

Best model based on test accuracy:
{best_model_name}

High-risk patient prediction:
{"Diabetes" if high_prediction == 1 else "No Diabetes"}

Low-risk patient prediction:
{"Diabetes" if low_prediction == 1 else "No Diabetes"}

Important note:
These predictions are machine-learning outputs for an educational project.
They are not medical diagnoses.
""")
```
